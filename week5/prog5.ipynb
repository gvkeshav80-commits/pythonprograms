```python
#tuple operations
t1=(12,3,4)
t2=(1,2,"hello","age")
print(t1+t2)
print(t2*2)
print(t2[2])
print("hello" in t2)
for i in t1:
  print(i)
print(len(t2))
```

```python
#tuple methods
t3=(14,15,7,7,8,"name","height")
print(t3.count(7))
print(t3.index("name"))
```

```python
#zip() function
t5=("n1","n2","n3")
t6=(1,23,34)
print(tuple(zip(t5,t6)))
```

```python
#concatenation of two tuples
s1=("veeraj",19,"vizag","geetham")
s2=("arjun",19,"bhimavaram","vit")
print(s1+s2)
```
