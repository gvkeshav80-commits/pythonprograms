```python
#to the number of duplicates in list
l=[int(i) for i in input().split()]
s=set(l)
print(s)
print(len(l)-len(s))
```

```python
s="python programs"
s1=set(s)
print(s1)
count = 0
for i in s1:
    if s.count(i) > 1:
        count += 1
print(count)
```

```python
#how many numbers having dupilactes
l=[int(i) for i in input().split()]
l2=[]
for i in l:
  if l.count(i)>1:
    l2.append(i)
print(l2)
s=set(l2)
print(s)
print(len(set(l2)))
```

```python
n=int(input())
l=[int(i) for i in input().split()]
l1=set(l)
elements=set(range(1,n+1))
print(elements)
print(elements-l1)
print(len(elements-l1))
```

```python
#methods in sets
a={1,2,3,"new"}
b={4,5,6,"world"}
a.add(6)
print(a)#add
print(a.union(b))#union
print(a&b)#intersection
print(a.difference(b))#difference
print(a.symmetric_difference(b))#symmetric_difference
b.update([1,2,3])#update
print(b)
a.remove(2)#remove
print(a)
b.discard(40)#discard
print(b)
a.clear()#clear
print(a)
b.pop()#pop
print(b)
b.pop()
print(b)

```
