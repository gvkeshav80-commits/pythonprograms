Dictionaries in python

```python
#built in functions
student={"name":"maharaj","branch":"ai&ds","year":2}
print(student)
print(student['name'])
student['age']=19
print(student)
student=dict(name="maharaj",branch="ai&ds",year=2)
print(student)
```

```python
k=[int(i) for i in input().split()]
v=[int(i) for i in input().split()]
d={i:j for i,j in zip(k,v)}
print(d)
n=int(input())
if n in d:
  print("yes")
else:
  print("no")
```

```python
l=[int(i) for i in input().split()]
d={i:l.count(i) for i in l}
print(d)
```
