```python
#list operations
l1=[1,25,36,"hello",7]
l2=[4,5,6,7,1]
print(l1+l2)#concatenation
print(l1*2)#repetition
print(l1[0])#indexing
print(l1[0:5])#slicing
print(l1[::2])
print(l2[:5])
print(2 not in l1)#membership
print(1 in l1)
print("hello" in l2)
#iteration
for i in l1:
    print(i)
for i in range(len(l1)):
    print(l1[i])
```

```python
#bulit in functions
l3=[2,36,28,30]
print(len(l3))
print(max(l3))
print(min(l3))
print(sum(l3))
print(sorted(l3))
print(list(reversed(l3)))

```

```python
#list methods
l4=[1,45,43,66,700]
l4.append(1)
print(l4)
l4.extend([2,3,400])
print(l4)
l4.insert(45,60)
print(l4)
l4.pop()
print(l4)
print(l4.count(3))
```

```python
#program that prints sum of list elements in list without using sum()
l5=[2,45,3,1]
sum=0
for(i) in l5:
    sum=sum+i
print(sum)
```

```python
#list compression
l=[i**2 for i in range(4)]
print(l)
```
